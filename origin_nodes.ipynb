```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np
import pandas as pd
import cartopy.crs as ccrs
import geopandas as gpd
from shapely.geometry import Polygon, MultiPolygon, Point
from shapely.ops import transform
import cartopy.feature as cfeature
import matplotlib.patheffects as pe
import textwrap
import matplotlib.colors as colors
import seaborn as sns
```

```python
# load origin node file
ds = pd.read_csv("datasets/data_viz/TC+TD_origin_node_count_perSubbasin_SyCLoPS.csv")

# fitler to Gulf only
ds = ds[['year', 'Gulf (A)', 'Gulf (B)']]

print(ds)
```

```python
# plot scatter plot
plt.plot(ds["year"], ds["Gulf (A)"], marker="o", label="Gulf (A)")
plt.plot(ds["year"], ds["Gulf (B)"], marker="o", label="Gulf (B)")

plt.xlabel("Year")
plt.ylabel("Value")
plt.title("Gulf (A) and Gulf (B) TC+TD Origins by Year")
plt.legend()

plt.savefig("images/data_viz/TC_origin/TC+TD/Gulf_AvsB_originNodes_perYr.png")
plt.show()
```

```python
# compute correlation between Gulf origin nodes
correlation = ds["Gulf (A)"].corr(ds["Gulf (B)"])

print(correlation)
```

```python
# scatter plot of Gulf A by Gulf B
plt.scatter(ds["Gulf (A)"], ds["Gulf (B)"])

plt.xlabel("Gulf (A)")
plt.ylabel("Gulf (B)")
plt.title(f"Gulf (A) vs Gulf (B) TC+TD Origins (r = {correlation:.2f})")

plt.savefig("images/data_viz/TC_origin/TC+TD/Gulf_AvsB_originNodes_correlation.png")
plt.show()
```
